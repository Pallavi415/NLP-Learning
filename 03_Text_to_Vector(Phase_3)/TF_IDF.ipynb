# TF-IDF

## Introduction

In this notebook, I will learn how **TF-IDF (Term Frequency-Inverse Document Frequency)** converts text into numerical vectors. Unlike simple Bag of Words, TF-IDF gives more importance to words that are frequent in a particular document but less common across the complete collection of documents.

## What is Term Frequency (TF)?

Term Frequency (TF) measures how frequently a word appears in a particular document. A word that appears many times in a document will have a higher TF value.

The basic formula is:

**TF = Number of times the word appears in a document / Total number of words in the document**

For example, if the word `good` appears 2 times in a document containing 10 words, its TF will be:

**TF = 2 / 10 = 0.2**

Therefore, TF tells me how important a word is within a particular document based on its frequency.

## TF Implementation

I will calculate Term Frequency manually first so that I can understand how the calculation works before using the TF-IDF Vectorizer.

```python
from collections import Counter
document = "good boy good girl"
words = document.split()
word_count = Counter(words)
total_words = len(words)
for word, count in word_count.items():
    tf = count / total_words
    print(word, "-->", tf)
```

## IDF Formula

The basic idea of IDF is:

**IDF = log(Total number of documents / Number of documents containing the word)**

A word that appears in almost every document will have a low IDF value, while a word that appears in fewer documents will have a higher IDF value.

```python
import math
from collections import Counter
corpus = [
    "good boy",
    "good girl",
    "good student"
]
total_documents = len(corpus)
words = set()
for document in corpus:
    words.update(document.split())
for word in words:
    document_count = sum(word in document.split() for document in corpus)

    idf = math.log(total_documents / document_count)
    print(word, "→", idf)
```

## Understanding IDF

If a word appears in **every document**, its IDF becomes `0` because that word does not help me distinguish between documents.

If a word appears in **only one document**, its IDF is higher because that word is more useful for identifying that particular document.

Therefore:

**Common word across documents → Low IDF**

**Rare word across documents → High IDF**

## TF-IDF

TF-IDF combines Term Frequency (TF) and Inverse Document Frequency (IDF) to calculate how important a word is in a particular document.

The formula is:

**TF-IDF = TF × IDF**

TF tells me how frequently a word appears in a document, while IDF tells me how rare the word is across all documents.

Therefore, a word gets a higher TF-IDF score when it appears frequently in a particular document but does not appear frequently in other documents.

## TF-IDF Implementation

In this implementation, I will calculate TF-IDF by multiplying the TF value of a word by its IDF value. This will help me understand how TF and IDF work together before using the `TfidfVectorizer`.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
corpus = [
    "He is a good boy",
    "She is a good girl",
    "He and She are good"
]
vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(corpus)
print(X.toarray())
```

## Vocabulary

I can use `get_feature_names_out()` to see the words that have been converted into features.

## TF-IDF Matrix

The TF-IDF matrix contains the numerical representation of my documents. Each row represents a document, each column represents a word, and each value represents the TF-IDF score of that word.

```python
import pandas as pd
tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)
tfidf_df
```
