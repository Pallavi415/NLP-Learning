# N-Grams

## What are N-Grams?

N-Grams are a sequence of `N` consecutive words taken from a sentence or text. They are used in NLP to capture the relationship between words that appear next to each other. The value of `N` determines how many words are included in each sequence.

For example, for the sentence:

`I love learning NLP`

- **Unigram (N=1):** `I`, `love`, `learning`, `NLP`
- **Bigram (N=2):** `I love`, `love learning`, `learning NLP`
- **Trigram (N=3):** `I love learning`, `love learning NLP`

N-Grams can provide some information about word order, unlike Bag of Words, which completely ignores word order.

## N-Grams Implementation

In this implementation, I will use NLTK to generate unigrams, bigrams, and trigrams from a sentence.

```python
from nltk import word_tokenize
```

```python
from nltk.util import ngrams
```

```python
sentence = "I love learning NLP"
```

```python
words = word_tokenize(sentence)
```

```python
print("Words:", words)
```

```python
#Unigrams
unigrams = list(ngrams(words, 1))
print("Unigrams:")
print(unigrams)
```

```python
#Bigrams
bigrams = list(ngrams(words, 2))
print("Bigrams:")
print(bigrams)
```

```python
#Trigrams
trigrams = list(ngrams(words, 3))
print("Trigrams:")
print(trigrams)
```

## Advantages of N-Grams

- **Captures word order:** N-Grams consider the order in which words appear.

- **Captures word combinations:** They can identify common combinations of words, such as `machine learning`.

- **Simple to understand:** N-Grams are easy to create and understand.

- **Useful for NLP tasks:** They can be used in text classification, language modeling, and text prediction.

- **Better than Bag of Words for word relationships:** N-Grams can capture some local relationships between neighboring words.

## Disadvantages of N-Grams

- **High dimensionality:** As the value of `N` increases, the number of possible word combinations can become very large.

- **Sparse representation:** Most possible N-Grams do not appear in a particular document, resulting in many zero values.

- **Does not understand semantic meaning:** N-Grams capture word combinations but do not truly understand the meaning of the words.

- **Limited context:** An N-Gram only looks at a fixed number of consecutive words. For example, a bigram only considers two words at a time.

- **Data sparsity:** Some meaningful N-Grams may not appear in the training data, making it difficult for a model to learn them.

## N-Grams vs Bag of Words

Bag of Words considers individual words and their frequency but ignores word order.

N-Grams consider groups of consecutive words, so they can capture some information about word order.

For example:

`I love NLP`

Bag of Words:

`I`, `love`, `NLP`

Bigram:

`I love`, `love NLP`

Therefore, **N-Grams provide more information about word order than Bag of Words.**
