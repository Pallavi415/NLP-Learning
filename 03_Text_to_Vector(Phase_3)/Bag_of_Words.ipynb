# Bag of Words

## What is Bag of Words?

Bag of Words (BoW) is a text representation technique used in NLP to convert text into numerical vectors. It represents a document based on the words present in it and their frequency. It is called a "bag" of words because it does not consider the order of the words in the sentence; it only focuses on which words are present and how many times they occur.

## Dataset

For this implementation, I will use three simple sentences as our dataset. I will first perform basic text preprocessing and then apply the Bag of Words technique.

```python
corpus = [
    "He is a good boy.",
    "She is a good girl.",
    "He and She are good."
]
```

```python
for sentence in corpus:
    print(sentence)
```

## Text Preprocessing

Before applying Bag of Words, we need to preprocess the text. In this implementation, I will convert the text to lowercase, remove punctuation, and remove common stop words. This helps us create a cleaner and more consistent vocabulary.

```python
import nltk
corpus = [
    "He is a good boy.",
    "She is a good girl.",
    "He and She are good."
]
print(corpus)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string
stop_words = set(stopwords.words("english"))
processed_corpus = []
for sentence in corpus:
    sentence = sentence.lower()
    words = word_tokenize(sentence)
    words = [
        word for word in words
        if word not in stop_words and word not in string.punctuation
    ]
    processed_corpus.append(words)
print(processed_corpus)
```

```python
## CREATE VOCABULARY
vocabulary = []
for sentence in processed_corpus:
    for word in sentence:
        if word not in vocabulary:
            vocabulary.append(word)
print(vocabulary)
```

```python
## CALCULATE FREQUENCY
from collections import Counter
all_words = []
for sentence in processed_corpus:
    all_words.extend(sentence)
frequency = Counter(all_words)
print(frequency)
```

```python
## SORT FREQUENCY - HIGH TO LOW  
frequency = frequency.most_common()
print(frequency)
```

```python
## CREATE BAG OF WORDS VECTOR
vocabulary = [word for word, count in frequency]
for sentence in processed_corpus:
    vector = []
    for word in vocabulary:
        vector.append(sentence.count(word))
    print(vector)
```

## Bag of Words Matrix

The Bag of Words vectors for all the sentences can be combined to form a **matrix**. Each row represents one sentence or document, and each column represents a word from the vocabulary. The values in the matrix represent how many times each word appears in the corresponding sentence.

For example, if our vocabulary is:

`['good', 'boy', 'girl']`

then the Bag of Words matrix is:

| Sentence | good | boy | girl |
|----------|------|-----|------|
| Sentence 1 | 1 | 1 | 0 |
| Sentence 2 | 1 | 0 | 1 |
| Sentence 3 | 1 | 0 | 0 |

This matrix is the numerical representation of our text that can be given to a machine learning model.

## Sparse Matrix

A Bag of Words representation usually produces a **sparse matrix**. A sparse matrix is a matrix in which most of the values are `0`. This happens because each sentence normally contains only a small number of words from the complete vocabulary.

For example:

`[1, 1, 0, 0, 0, 0, 0, 0]`

contains mostly zeros, so it is called a sparse representation.

Sparse matrices are useful because they can store large text datasets more efficiently without unnecessarily storing all the zero values.

## Advantages of Bag of Words

- **Simple and easy to understand:** Bag of Words is one of the simplest techniques for converting text into numerical form.

- **Easy to implement:** It can be implemented using simple Python code or NLP libraries.

- **Captures word frequency:** It tells us how many times each word appears in a sentence or document.

- **Works well with machine learning:** The numerical vectors can be directly used with traditional machine learning algorithms.

- **Useful for simple NLP tasks:** It can be useful for tasks such as spam detection, document classification, and basic sentiment analysis.

## Disadvantages of Bag of Words

- **High dimensionality:** If the dataset contains thousands of unique words, the vocabulary becomes very large and the vectors also become very large.

- **Sparse representation:** Most values in the vectors are `0`, which can make the representation inefficient for large vocabularies.

- **Ignores word order:** Bag of Words does not consider the order of words in a sentence.

For example:

`Dog bites man`

and

`Man bites dog`

contain the same words, but their meanings are different. Bag of Words may represent them similarly because it mainly considers word occurrence.

- **Does not understand semantic meaning:** Bag of Words does not understand the actual meaning or relationship between words.
