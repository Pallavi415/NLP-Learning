# Stop Words

## What are Stop Words?

Stop words are commonly used words that usually do not provide much useful information for text analysis. Examples of stop words include `the`, `is`, `a`, `an`, `and`, `in`, `of`, `to`, and `for`. In NLP, stop words are often removed during text preprocessing to reduce unnecessary words and make the text easier and faster to process.

## Why Remove Stop Words?

Removing stop words can help reduce the number of words that need to be processed. This can make NLP models faster and can help focus on the more important words in a sentence. However, stop words should not always be removed because some NLP tasks may need them to understand the meaning of a sentence.

## Stop Words Implementation

In this implementation, we will use the `stopwords` corpus from NLTK to identify commonly used English stop words. We will first download the required NLTK resource and then check whether the words in a sentence are stop words.


```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
print(stop_words)
```

```python
##Check Stop Words in a Sentence
sentence = "I am learning Natural Language Processing"
words = sentence.lower().split()
for word in words:
    if word in stop_words:
        print(word, "---> Stop Word")
    else:
        print(word, "--> Not a Stop Word")
```

```python
##Remove Stop Words
sentence = "I am learning Natural Language Processing"
words = sentence.lower().split()
filtered_words = []
for word in words:
    if word not in stop_words:
        filtered_words.append(word)
print("Original Sentence:", sentence)
print("After Removing Stop Words:", filtered_words)
```
