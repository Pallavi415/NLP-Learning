# Stemming

## What is Stemming?

Stemming is a text preprocessing technique in NLP that reduces a word to its root or base form by removing prefixes or suffixes. The main purpose of stemming is to group different forms of the same word together so that they can be treated as a single word during text analysis. For example, words such as `playing`, `played`, and `plays` can be reduced to `play`. Stemming is a simple and fast technique, but it may sometimes produce a word that is not a meaningful dictionary word.

## Stemming Implementation

In this implementation, we will use the `PorterStemmer` from NLTK to reduce words to their stem. We will apply the stemmer to a list of words and compare the original words with their stemmed forms.

```python
from nltk.stem import PorterStemmer
```

```python
##type of stemming
stemmer = PorterStemmer()
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "running",
    "runner",
    "easily",
    "fairly",
    "happily",
    "connected",
    "connecting",
    "connection",
    "education",
    "educational",
    "beautiful",
    "beautifully",
    "organized",
    "organizing",
    "organization",
    "reading",
    "reader",
    "reads",
    "writing",
    "writer",
    "written",
    "jumping",
    "jumped",
    "jumps",
    "fishing",
    "fished",
    "fisher"
]
```

```python
for word in words:
    print(word+"----->"+stemmer.stem(word))
```

## Disadvantages of Stemming

- **Can produce meaningless words:** Stemming may create words that are not valid dictionary words. For example, `studies` may become `studi`.

- **Can remove too much:** Sometimes stemming removes important parts of a word and changes its meaning.

- **Can produce incorrect roots:** Different words may be reduced to the same stem even though their meanings are different.

- **Less accurate than lemmatization:** Stemming mainly removes word endings without understanding the actual meaning or context of the word.

- **Not suitable for every NLP task:** Tasks that require accurate meaning and language understanding may benefit more from lemmatization than stemming.

# Regexp Stemmer

## What is Regexp Stemmer?

Regexp Stemmer is a stemming technique in NLP that uses **regular expressions (regex)** to remove specific patterns or endings from words. Unlike Porter or Lancaster Stemmer, where predefined stemming rules are used, we can define our own rules according to the requirement. It is useful when we want more control over which parts of a word should be removed.

```python
from nltk.stem import RegexpStemmer
```

```python
stemmer = RegexpStemmer(r'ing$|s$|ed$', min=4)
```

```python
words = [
    "playing",
    "played",
    "plays",
    "running",
    "jumping",
    "jumped",
    "jumps",
    "reading",
    "reads",
    "talking",
    "talked",
    "talks"
]
```

```python
for word in words:
    print(word, "---->", stemmer.stem(word))
```

## Understanding the Regular Expression

The regular expression `r'ing$|s$|ed$'` tells the stemmer to remove specific endings from a word.

- `ing$` → removes `ing` from the end of a word.
- `s$` → removes `s` from the end of a word.
- `ed$` → removes `ed` from the end of a word.
- `|` → means **OR**.
- `$` → means the pattern must occur at the **end of the word**.
- `min=4` → prevents the stemmer from reducing words below 4 characters.

# Snowball Stemmer

## What is Snowball Stemmer?

Snowball Stemmer is a stemming technique in NLP that reduces words to their root or base form by removing common prefixes and suffixes. It is an improved and more flexible stemming algorithm compared to the Porter Stemmer. Snowball Stemmer supports multiple languages and is commonly used for text preprocessing in NLP.

```python
from nltk.stem import SnowballStemmer
stemmer = SnowballStemmer("english")
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "running",
    "runner",
    "easily",
    "fairly",
    "connected",
    "connecting",
    "connection",
    "education",
    "educational",
    "beautiful",
    "beautifully"
]
for word in words:
    print(word, "--->", stemmer.stem(word))
```

## Snowball Stemmer Performance

Snowball Stemmer generally provides good stemming results while being relatively fast and lightweight. It is often preferred over the original Porter Stemmer because its algorithms are more refined. However, like other stemming techniques, it may sometimes produce stems that are not actual dictionary words.
