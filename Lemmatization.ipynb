# Lemmatization

## What is Lemmatization?

Lemmatization is an NLP preprocessing technique that reduces a word to its **base or dictionary form**, called a **lemma**. It considers the meaning and grammatical form of the word to find the correct base form. For example, `running` becomes `run`, `studies` becomes `study`, and `better` can become `good`. Unlike stemming, lemmatization usually produces a meaningful dictionary word.

## Lemmatization Implementation

In this implementation, we will use `WordNetLemmatizer` from NLTK to convert words into their meaningful base or dictionary form. We will apply the lemmatizer to a list of words and compare the original words with their lemmas.

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
### WordNetLemmatizer
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "running",
    "ran",
    "runs",
    "children",
    "child",
    "cars",
    "mice",
    "geese",
    "feet",
    "better",
    "best",
    "worse",
    "worst",
    "eating",
    "ate",
    "eaten",
    "writing",
    "wrote",
    "written",
    "reading",
    "read",
    "talking",
    "talked",
    "talks",
    "walking",
    "walked",
    "walking",
    "driving",
    "drove",
    "driven",
    "buying",
    "bought",
    "selling",
    "sold",
    "learning",
    "learned",
    "working",
    "worked"
]
```

```python
for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
```

# POS Tagging

## What is POS Tagging?

POS (Part-of-Speech) Tagging is an NLP technique used to identify the **grammatical category of each word** in a sentence. It tells us whether a word is a noun, verb, adjective, adverb, pronoun, etc.

For example, in the sentence `I am learning NLP`, `I` is a pronoun, `am` is a verb, `learning` is a verb, and `NLP` is a noun.

POS tagging is useful for understanding the grammatical structure of a sentence and is commonly used in tasks such as text analysis, information extraction, and language understanding.

## Common POS Tags

- **NN** → Noun
- **VB** → Verb
- **JJ** → Adjective
- **RB** → Adverb
- **PRP** → Personal Pronoun
- **DT** → Determiner
- **IN** → Preposition
- **CC** → Conjunction

```python
for word in words:
    print(word, "→", lemmatizer.lemmatize(word, pos="v"))
```
